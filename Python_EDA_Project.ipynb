```python
# Function to identify outliers using the IQR method

def find_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]
    
    return outliers, lower_bound, upper_bound
```

```python
# Detect outliers for important numerical variables

outlier_columns = [
    "average_cost_for_two",
    "aggregate_rating",
    "votes",
    "photo_count"
]

for column in outlier_columns:
    
    outliers, lower, upper = find_outliers(df_clean, column)
    
    print(f"\n{column}")
    print("-" * 50)
    print("Lower Bound:", round(lower, 2))
    print("Upper Bound:", round(upper, 2))
    print("Number of Outliers:", len(outliers))
```

```python
# Define broad Indian regions

north_cities = [
    "Agra", "Amritsar", "Chandigarh", "Delhi",
    "Ghaziabad", "Gurgaon", "Jaipur", "Lucknow",
    "Ludhiana", "Noida", "Patna", "Varanasi"
]

west_cities = [
    "Ahmedabad", "Goa", "Mumbai", "Nashik",
    "Pune", "Surat", "Vadodara"
]

south_cities = [
    "Bangalore", "Bengaluru", "Chennai", "Coimbatore",
    "Hyderabad", "Kochi", "Mangalore", "Mysore",
    "Thiruvananthapuram"
]

east_cities = [
    "Bhubaneshwar", "Bhubaneswar", "Kolkata",
    "Ranchi"
]
```

```python
def assign_region(city):
    
    if city in north_cities:
        return "North"
    
    elif city in west_cities:
        return "West"
    
    elif city in south_cities:
        return "South"
    
    elif city in east_cities:
        return "East"
    
    else:
        return "Other"
```

```python
df_clean["region"] = df_clean["city"].apply(assign_region)

df_clean[["city", "region"]].head(20)
```
